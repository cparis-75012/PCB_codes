```python
# calls "GOOD CELLS" those whose MEAN SQUARE ERROR of the FTVM fit is LESS THAN 10

import numpy as np
import import_ipynb
import ipynb.fs.full.ssiProjectModule as ssiMod

import matplotlib as mpl
mpl.rcParams['axes.spines.right'] = False
mpl.rcParams['axes.spines.top'] = False

import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt

import os
import pickle

props = {'boxstyle':'round', 'edgecolor':'white', 'facecolor':'white', 'alpha':0}
main_dir = 'C:/Users/Carlo PARIS/Documents/Career/PhD/Science/Ferrari/Data/DS Cells - Franke/'
stim = np.arange(0, 360, 10)
total_rejected = 0
threshold_error = 15

for experiment_number in range(1,8,1):
    experiment_name = f'Exp{experiment_number}'
    datanfits = dict(np.load(f'{main_dir+experiment_name}/Analysis of {experiment_name}.npz'))
    good_cells = np.unique(np.where(datanfits['mse_ftvm_fit'][:,1]<threshold_error)[0])
    bad_cells = np.unique(np.where(datanfits['mse_ftvm_fit'][:,1]>=threshold_error)[0])
    
    datanfits['good_cell_indeces'] = good_cells
    print(f'Good cells in {experiment_name}',datanfits['good_cell_indeces'])
    np.savez(f'{main_dir+experiment_name}/Analysis of {experiment_name}.npz', **datanfits)

    if len(bad_cells)==0:
        print(f'No cells were rejected from {experiment_name}', end='\n\n')
    else:
        print(f'Bad cells in {experiment_name}:{bad_cells}', end='\n\n')
    total_rejected += len(bad_cells)

print('Total number of rejected cells: ', total_rejected)
ssiMod.COPY_CURRENT_NB(main_dir, 'Find Cells w Fit MSE Larger-Than 10')
```
